# Split Dependent (Target/Label) Independent (atribut) Variabel

```python
x = df_iris.iloc[:,[1, 2, 3, 4]].values
y = df_iris.iloc[:, [5]].values
x 
```

```python
y = y.reshape(len(y),)
```

# Feature Scalling

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
x = scaler.fit_transform(x) # untuk menyamakan ukuran angka di setiap kolom data
```

# Split Train Test

```python
from sklearn.model_selection import train_test_split

# Skenario 1: Train 80% : Test 20%
x_train_1, x_test_1, y_train_1, y_test_1 = train_test_split(x, y, test_size=0.2, random_state=0)

# Skenario 2: Train 70% : Test 30%
x_train_2, x_test_2, y_train_2, y_test_2 = train_test_split(x, y, test_size=0.3, random_state=0)

# Skenario 3: Train 60% : Test 40%
x_train_3, x_test_3, y_train_3, y_test_3 = train_test_split(x, y, test_size=0.4, random_state=0)
```

# Klasifikasi KNN

```python
from sklearn.neighbors import KNeighborsClassifier
```

# Training/Fit Model

```python
knn_1 = KNeighborsClassifier(n_neighbors=5, metric='euclidean').fit(x_train_1, y_train_1)
knn_1

knn_2 = KNeighborsClassifier(n_neighbors=1, metric='euclidean').fit(x_train_2, y_train_2)
knn_2

knn_3 = KNeighborsClassifier(n_neighbors=1, metric='euclidean').fit(x_train_3, y_train_3)
knn_3
```

# Prediksi/Test

```python
y_prediksi1 = knn_1.predict(x_test_1)
y_prediksi2 = knn_2.predict(x_test_2)
y_prediksi3 = knn_3.predict(x_test_3)
```

```python
print("Prediksi Skenario 1:", y_prediksi1)
print("Prediksi Skenario 2:", y_prediksi2)
print("Prediksi Skenario 3:", y_prediksi3)
```

# Cek Akurasi dan Evaluasi

```python
import sys
!{sys.executable} -m pip install seaborn --break-system-packages
```

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
print("Nilai Akurasi Skenario 1:", accuracy_score(y_test_1, y_prediksi1))
print("Nilai Akurasi Skenario 2:", accuracy_score(y_test_2, y_prediksi2))
print("Nilai Akurasi Skenario 3:", accuracy_score(y_test_3, y_prediksi3))
```

```python
print("\nClassification Report Skenario 1:")
print(classification_report(y_test_1, y_prediksi1))
sb.heatmap(confusion_matrix(y_test_1, y_prediksi1), annot=True, cmap='Blues')
plt.title("Confusion Matrix - Skenario 1")
plt.show()

print("\nClassification Report Skenario 2:")
print(classification_report(y_test_2, y_prediksi2))
sb.heatmap(confusion_matrix(y_test_2, y_prediksi2), annot=True, cmap='Greens')
plt.title("Confusion Matrix - Skenario 2")
plt.show()

print("\nClassification Report Skenario 3:")
print(classification_report(y_test_3, y_prediksi3))
sb.heatmap(confusion_matrix(y_test_3, y_prediksi3), annot=True, cmap='Oranges')
plt.title("Confusion Matrix - Skenario 3")
plt.show()
```

# Mencari NIlai K yang Terbaik

```python
n_akurasi_1 = []
for k in range (1, 21):
    knn = KNeighborsClassifier(n_neighbors=k,metric='euclidean').fit(x_train_1, y_train_1)
    y_prediksi = knn.predict(x_test_1)
    akurasi = accuracy_score(y_test_1, y_prediksi)
    n_akurasi_1.append(akurasi)
print("Akurasi Skenario 1 (K=1 sampai 21):")
print(n_akurasi_1)
print(f"K Terbaik Skenario 1 adalah {n_akurasi_1.index(max(n_akurasi_1)) + 1} dengan akurasi {max(n_akurasi_1)}\n")

n_akurasi_2 = []
for k in range(1, 21):
    knn = KNeighborsClassifier(n_neighbors=k, metric='euclidean').fit(x_train_2, y_train_2)
    y_prediksi = knn.predict(x_test_2)
    akurasi = accuracy_score(y_test_2, y_prediksi)
    n_akurasi_2.append(akurasi)
print("Akurasi Skenario 2 (K=1 sampai 21):")
print(n_akurasi_2)
print(f"K Terbaik Skenario 2 adalah {n_akurasi_2.index(max(n_akurasi_2)) + 1} dengan akurasi {max(n_akurasi_2)}\n")

n_akurasi_3 = []
for k in range(1, 21):
    knn = KNeighborsClassifier(n_neighbors=k, metric='euclidean').fit(x_train_3, y_train_3)
    y_prediksi = knn.predict(x_test_3)
    akurasi = accuracy_score(y_test_3, y_prediksi)
    n_akurasi_3.append(akurasi)
print("Akurasi Skenario 3 (K=1 sampai 21):")
print(n_akurasi_3)
print(f"K Terbaik Skenario 3 adalah {n_akurasi_3.index(max(n_akurasi_3)) + 1} dengan akurasi {max(n_akurasi_3)}")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))
plt.plot(range(1, 21), n_akurasi_1)
plt.title("Grafik Akurasi Skenario 1 (80:20)")
plt.xlabel("Nilai K")
plt.ylabel("Akurasi")
plt.show()

plt.figure(figsize=(20, 10))
plt.plot(range(1, 21), n_akurasi_2)
plt.title("Grafik Akurasi Skenario 2 (70:30)")
plt.xlabel("Nilai K")
plt.ylabel("Akurasi")
plt.show()

import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))
plt.plot(range(1, 21), n_akurasi_3)
plt.title("Grafik Akurasi Skenario 3 (60:40)")
plt.xlabel("Nilai K")
plt.ylabel("Akurasi")
plt.show()

```

# Klasifikasi Data Baru

```python
# Urutan: SepalLength, SepalWidth, PetalLength, PetalWidth
data_baru = [[5.1, 3.5, 1.4, 0.2]]

# Wajib di-scale dulu pakai scaler yang sama
data_baru_scaled = scaler.transform(data_baru)

# Prediksi dari masing-masing skenario
prediksi_1 = knn_1.predict(data_baru_scaled)
prediksi_2 = knn_2.predict(data_baru_scaled)
prediksi_3 = knn_3.predict(data_baru_scaled)

print("Prediksi Skenario 1 (80:20):", prediksi_1)
print("Prediksi Skenario 2 (70:30):", prediksi_2)
print("Prediksi Skenario 3 (60:40):", prediksi_3)
```

# HASIL
